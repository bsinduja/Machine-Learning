```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv('Mall_Customers.csv')
```

```python
dataset
```

```python
X = dataset.iloc[:, 2:5].values
```

```python
X
```

```python
from sklearn.neighbors import NearestNeighbors
import numpy as np
import matplotlib.pyplot as plt
#from kneed import KneeLocator

# Assuming 'X_scaled' is your standardized feature set (highly recommended for DBSCAN)
# Replace 'X_scaled' with your actual data variable

# Compute distances to the MinPts-th nearest neighbor (k = MinPts)
neighbors = NearestNeighbors(n_neighbors=10) # use 10 as min_samples
neighbors_fit = neighbors.fit(X)
distances, indices = neighbors_fit.kneighbors(X)

# Sort distances and plot
distances = np.sort(distances[:, 9], axis=0) # Index 9 corresponds to the 10th neighbor (0-indexed)
plt.plot(distances)
plt.xlabel("Points sorted by distance")
plt.ylabel("Epsilon (eps)")
plt.title("Choosing the Right eps using k-distance Graph")
plt.show()

# Alternatively, use the kneed library to find the elbow automatically
# knee_locator = KneeLocator(range(len(distances)), distances, curve="concave", direction="increasing")
# optimal_eps = knee_locator.knee 
```

```python
from sklearn.cluster import DBSCAN
clusmodel = DBSCAN(eps=15,min_samples=10).fit(X)
label = clusmodel.labels_
```

```python
label
```

```python
supervised=dataset
```

```python
supervised
```

```python
supervised['Cluster_group']=label
```
