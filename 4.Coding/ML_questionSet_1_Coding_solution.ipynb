```python
import pandas as pd

data = {'Customer_ID': [101, 102, 103, 101, 104, 102, 101, 105, 102, 103],
        'Purchase_Amount': [200, 150, 180, 220, 300, 200, 100, 400, 250, 300]}

df=pd.DataFrame(data)

per_customer_purchase=df.groupby('Customer_ID')['Purchase_Amount'].sum()

print(per_customer_purchase)
print(per_customer_purchase.sort_values(ascending=False).head(3))
```

```python
from sklearn.preprocessing import StandardScaler

dataset = {'Square_Feet': [1500, 2000, 2500, 3000, 3500],
        'Price': [300000, 400000, 500000, 600000, 700000]}
df=pd.DataFrame(dataset)
ind_var=ind_var=df[["Square_Feet"]]
dep_var=df[["Price"]]

#ind_var
#dep_var

#import matplotlib.pyplot as plt 
#plt.scatter(ind_var,dep_var)
#plt.xlabel("SqFt")
#plt.ylabel("Price")
#plt.show()

from sklearn.model_selection import train_test_split 
X_train,X_test,Y_train,Y_test=train_test_split(ind_var,dep_var,test_size=0.30,random_state=0)
from sklearn.linear_model import LinearRegression 
regressor=LinearRegression()
regressor.fit(X_train,Y_train)
Y_pred=regressor.predict(X_test)
from sklearn.metrics import r2_score
r_score=r2_score(Y_test,Y_pred)
#r_score

house_price=regressor.predict([[1800], [2800]])

print("Predicted Prices:\n",house_price)
#house_price
```

```python
from collections import Counter
import pandas as pd
data = {'Category': ['A', 'B', 'C', 'A', 'D', 'B', 'A', 'E', 'B', 'C', 'C', 'A'],
        'Value': [10, 15, 20, 30, 25, 18, 22, 40, 35, 50, 45, 15]}
df=pd.DataFrame(data)

counts=df.groupby('Category').size().reset_index(name='Count')

print('Total Occurrences per Category:\n',counts)

print('Top 3 Frequent Categories:\n',(counts.sort_values(ascending=False,by='Count').head(3)))

#counts=df.stack().value_counts()
#counts
```
