```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv('Mall_Customers.csv')
```

```python
dataset
```

```python
X = dataset.iloc[:, 2:5].values
```

```python
X
```

```python
from sklearn.cluster import MeanShift, estimate_bandwidth
bandwidth = estimate_bandwidth(X, quantile=0.1, n_samples=500)
#when quantile was 0.2 all datapoints fell into single cluster. when 0.05 and sample=500 13 clusters formed. when sample was 20 and quantile was 0.05 42 clsuters formed.
clusmodel = MeanShift(bandwidth=bandwidth,bin_seeding=True)
label = clusmodel.fit_predict(X)
```

```python
bandwidth 
```

```python
label
```

```python
supervised=dataset
```

```python
supervised['Cluster_group']=label
```

```python
supervised
```

```python
supervised.to_csv("meanshift.csv",index=False)
```
