```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv('Mall_Customers.csv')
```

```python
dataset
```

```python
X = dataset.iloc[:, 2:5].values
```

```python
X
```

```python
from sklearn.cluster import OPTICS, cluster_optics_dbscan
clusmodel = OPTICS(min_samples=10, xi=0.05, min_cluster_size=0.05).fit(X)
label = clusmodel.labels_
```

```python
label
```

```python
supervised=dataset
```

```python
supervised
```

```python
supervised['Cluster_group']=label
```

```python
supervised
```

```python
supervised.to_csv("OPTICS.csv",index=False)
```
